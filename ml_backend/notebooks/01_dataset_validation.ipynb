# Dataset Validation
This notebook verifies that all images referenced in `metadata.csv` exist and are uncorrupted.

```python
import os
import pandas as pd
import cv2
from tqdm import tqdm

metadata_path = '../data/processed/metadata.csv'
if not os.path.exists(metadata_path):
    print('Metadata not found. Please run the data ingestion pipeline first.')
else:
    df = pd.read_csv(metadata_path)
    print(f'Total images in metadata: {len(df)}')
    
    missing = 0
    corrupted = 0
    
    for path in tqdm(df['image_path'], desc='Validating Images'):
        # Adjust path relative to notebook location if needed
        full_path = os.path.join('../', path) 
        if not os.path.exists(full_path):
            missing += 1
        else:
            img = cv2.imread(full_path)
            if img is None:
                corrupted += 1
                
    print(f'\nMissing files: {missing}')
    print(f'Corrupted files: {corrupted}')
    if missing == 0 and corrupted == 0:
        print('Dataset is 100% valid!')
```
