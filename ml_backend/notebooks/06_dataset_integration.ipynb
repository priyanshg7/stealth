# Automated Kaggle Dataset Integration
Executes the entire end-to-end pipeline to standardize the 20K Multi-class crop disease dataset into the KisanMitra format.

```python
import sys
import os
sys.path.append('../src/')
sys.path.append('../scripts/')

# 1. Download & Extract
from download_20k_dataset import setup_20k_dataset
setup_20k_dataset()
```

```python
# 2. Analyze Raw Dataset
from analyze_dataset import analyze_raw_dataset
analyze_raw_dataset('../data/raw/20k_dataset')
```

```python
# 3 & 4. Map Labels & Filter Supported Classes
from filter_classes import filter_dataset
filter_dataset('../data/raw/20k_dataset', '../data/raw/staged_dataset')
```

```python
# 5. Image Validation (Corrupted, Grayscale, Low-Res)
from image_validator import validate_images
validate_images('../data/raw/staged_dataset', '../data/raw/validated_dataset', '../data/rejected/')
```

```python
# 6. Perceptual Deduplication
from deduplicate import deduplicate_dataset
deduplicate_dataset('../data/raw/validated_dataset')
```

```python
# 7. Standardize Images & Remove EXIF
from image_standardizer import standardize_images
image_records = standardize_images('../data/raw/validated_dataset', '../data/processed/images/')
```

```python
# 8. Generate Metadata
from metadata_generator import generate_metadata
metadata_df = generate_metadata(image_records)
print(f"Generated metadata for {len(metadata_df)} images.")
```

```python
# 9. Stratified Train/Val/Test Split
from train_split import generate_splits
final_df = generate_splits(metadata_df)
```

```python
# 10. Generate Final Report
from dataset_report import generate_report
generate_report(final_df)
```
